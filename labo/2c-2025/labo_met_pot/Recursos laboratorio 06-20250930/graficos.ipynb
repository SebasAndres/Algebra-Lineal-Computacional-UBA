```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import FancyArrowPatch
```

```python
np.random.seed(42)

A1 = np.array([[0.5,0.5],
               [0.5, 0.5]])

R = 5
K = [1,5,10,50,100]

vectoresAzar = [np.random.rand(2) for _ in range(R)]

colores = ['green', 'red','blue', 'black', 'yellow']

fig, ax = plt.subplots()

for i, k in enumerate(K):
    v =vectoresAzar[0]
    ax.add_patch(
        FancyArrowPatch(
            v,  # origen
            np.power(A1,k)@v/np.linalg.norm(np.power(A1,k)@v), # extremo
            arrowstyle='->',   # estilo con punta
            mutation_scale=15, # tamaño de la punta
            color=colores[i]
        )
    )

ax.set_aspect('equal') 
```
